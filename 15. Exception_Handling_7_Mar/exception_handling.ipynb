### modulo by zero

```python
try:
    reminder=17%0
except ZeroDivisionError as err:
    print("you cannot get remainder when dividing by zero please provide other value:-",err)
```

```python
def safe_div(a,b):
    return a/b if b!=0 else "undefined"

print(safe_div(10,2))
```

```python
print(safe_div(10,0))
```

### OverflowError

```python
import math
try:
    huge=math.exp(1000)
    print(huge)

except OverflowError as err:
    print("Overflow error caught:",err)
```

```python
big_int=10**500
```

```python
print(big_int)
```

```python
try:
    x=1.8e308*10
    print(x)
except OverflowError as err:
    print("Overflow error caught:",err)
```
